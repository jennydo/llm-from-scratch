```python
import torch
import numpy as np
import time
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(device)
```

```python
start_time = time.time()
# matrix operations here
zeros = torch.zeros(1, 1)
end_time = time.time()

elapse_time = end_time - start_time
print(f"{elapse_time:.4f}")
```

```python
# CPU vs GPU performance
# GPU is faster especially for matrix multiplication

torch_rand1 = torch.rand(100, 100, 100, 100).to(device)   # to device -> cuda (gpu)
torch_rand2 = torch.rand(100, 100, 100, 100).to(device)
np_rand1 = torch.rand(100, 100, 100, 100)  # cpu
np_rand2 = torch.rand(100, 100, 100, 100)

start_time = time.time()
rand = (torch_rand1 @ torch_rand2)  # matrix multiplication (multiply rows from the first matrix with columns from the second matrix)
end_time = time.time()
elapsed_time = end_time - start_time
print(f"{elapse_time:.8f}")


start_time = time.time()
# rand = np.multiply(np_rand1, np_rand2)  # element-wise multiplication (Multiply corresponding elements one-by-one at the same positions)
rand = np_rand1 @ np_rand2
end_time = time.time()
elapsed_time = end_time - start_time
print(f"{elapse_time:.8f}")
```
