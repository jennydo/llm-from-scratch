```python
# CPU vs GPU performance
# GPU is faster especially for matrix multiplication. Why?
# CPU: few powerful cores
# GPU: thousands of smaller parallel cores
# -> use GPU for training efficiency

# GPU tensors
torch_rand1 = torch.rand(10000, 10000).to(device)
torch_rand2 = torch.rand(10000, 10000).to(device)

# CPU numpy arrays
np_rand1 = np.random.rand(10000, 10000)
np_rand2 = np.random.rand(10000, 10000)

# GPU timing
torch.cuda.synchronize()   # to proper time GPU
start_time = time.time()
rand = torch.matmul(torch_rand1, torch_rand2)  # matrix multiplication 
torch.cuda.synchronize()
end_time = time.time()
elapsed_time = end_time - start_time
print(f"GPU time: {elapsed_time:.8f} seconds")

# CPU timing
start_time = time.time()
rand = np.dot(np_rand1, np_rand2) # dot product (similar to matrix multiplication)
end_time = time.time()
elapsed_time = end_time - start_time
print(f"CPU time: {elapsed_time:.8f} seconds")
```

```python
# torch.multinomial: randomly samples indices from a probability distribution:
# Define a probability tensor
probabilities = torch.tensor([0.1, 0.9])
# 10% or 0.1 => 0, 90% or 0.9 => 1. each probability points to the index of the probability in the tensor

# Draw 10 samples from the multinomial distribution
samples = torch.multinomial(probabilities, num_samples=10, replacement=True)
print(samples)
```

```python
# torch.cat: Concatenation
tensor = torch.tensor([1, 2, 3, 4])
out = torch.cat((tensor, torch.tensor([5])), dim=0)  # Join tensors together along dimension 0
out

tensor = torch.tensor([[1, 2],
                       [3, 4]])
out = torch.cat((tensor, torch.tensor([[5], [6]])), dim=1)  # Join tensors together along dimension 1
out

# Very common for:
# combining embeddings
# building sequences
# appending generated tokens
# batching data
```

```python
# creates a lower triangular matrix
out = torch.tril(torch.ones(5, 5))
out

# This exact operation is used for:
# causal masking
# autoregressive generation
# because GPT should NOT see future tokens
```

```python
# creates an upper triangular matrix
out = torch.triu(torch.ones(5, 5))
out

# causal masking
# future positions
# invalid attention regions
```

```python
# masked_fill() replaces elements in a tensor wherever a mask is True
out = torch.zeros(5, 5).masked_fill(torch.tril(torch.ones(5, 5)) == 0, float('-inf'))
out
# == 0: converts to boolean mask -> 1 = False, 0 = True. Replace with float('-inf') where it's True

# 1. Attention masking: Hide future tokens
# 2. Padding masks: Ignore padding tokens
# 3. Data cleaning: Replace invalid values
```

```python
# exp(x)= e^x for all entries in out
torch.exp(out)

# This is the core trick behind:
# softmax
# attention masking
# probability conversion
```

```python
# Neural Network
# y=Wx+b (linear)
# W: weight, b: bias
# y: output, x: input vector

import torch.nn as nn
sample = torch.tensor([10., 20., 15.])
linear = nn.Linear(3, 3, bias=False)   #in, out size = 3
print(linear.weight)   #y = Wx -> W is 3x3 matrix
print(linear(sample))

# https://docs.google.com/presentation/d/1xslguVqu0sHANOhJduP31GJLFX-LbjVGS1wfc7Jvy6A/edit?usp=sharing

# Input -> Hidden layers -> Output
# The node at each hidden layers are activation functions
# Activation functions:
# - Linear
# - Sigmoid
# - ReLU
# - Tanh

# The lines connecting layers are weights -> put it in matrix (eg y1=w11​x1+w12​x2+w13​x3)
# Training a neural network means: adjusting these weights so outputs become more correct
# Gradient descent updates them during learning

# Mean Squared Error (MSE) is a popular loss function for regression algorithms, 
# measuring the average squared difference between predicted and actual values
# Lower MSE indicates better model performance

# MLP Multilayer perceptrons (General purpose predictions)
# Transformers (LLM)
# RNN (recurrent neural networks) processing sequential data
# CNN: computer vision
```

```python
import torch.nn.functional as F

tensor1 = torch.tensor([1.0, 2.0, 3.0])
softmax_out = F.softmax(tensor1, dim=0)
print(softmax_out)

# Softmax: convert a vector of raw scores (logits) into a probability distribution, 
# where each value ranges between 0 and 1 and all values sum to 1
# https://en.wikipedia.org/wiki/Softmax_function to see the func

# Exponentials amplify differences -> amplifies larger values and reduces smaller values -> confident prediction
```

```python
# Embedding Vectors

# Initialize an embedding layer
vocab_size = 80
embedding_dim = 6
embedding = nn.Embedding(vocab_size, embedding_dim)

# Create some input indices
input_indices = torch.LongTensor([1, 5, 3, 2])  
# embeddings for: token 1, 5, 3, 2

# Apply the embedding layer
embedded_output = embedding(input_indices)

# The output will be a tensor of shape (4, 6), where 4 is the number of inputs and 6 is the dimensionality of the embedding vectors
print(embedded_output.shape)
print(embedded_output)


# Embedding
# representation learning that maps complex, high-dimensional data into a lower-dimensional vector
# Similar things get similar vectors
# Different things get distant vectors

# In LLM
# Text
# → Token IDs
# → Embeddings
# → Transformer layers
# → Predictions
```

```python
# Dot product
# a · b = a1b1 + a2b2 + a3b3

# a = [1, 2, 3]
# b = [4, 5, 6]

# dot = 1*4 + 2*5 + 3*6 = 4 + 10 + 18 = 32

# Matrix multiplication
# (row from first matrix) DOT (column from second matrix)

a = torch.tensor([[1,2],[3,4],[5,6]])   # 3x2
b = torch.tensor([[7,8,9],[10,11,12]])  # 2x3
# print(a @ b)   # 3x3 same as matmul
print(torch.matmul(a, b))
```
